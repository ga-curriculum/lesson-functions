## Review: Vowel Counter
Before getting into functions, let's review control flow and other operations.

````python
# Define a string to use.
input_string = "journey before destination"

# Instantiate counter at 0.
n_vowels = 0

# Iterate through our string.
# What do we need to look out for?
# Note that we use .lower() to account for upper-
# case and lower-case vowels!
for char in input_string.lower():
    
    # Check to see if the letter in our string
    # is a vowel.
    if char in 'aeiou':
        
        # If that's the case, then add 1 to our count.
        n_vowels += 1
        
# Print out the number of vowels.
print(n_vowels)
````

## Review: FizzBuzz
Let's tackle the FizzBuzz problem. (This is a common interview question!)

- Print out every number between 1 and 100, _inclusive_. But...
- If the number is divisible by 3, print "Fizz" instead of the number.
- If the number is divisible by 5, print "Buzz" instead of the number.
- If the number is divisible by 3 _and_ 5, print "FizzBuzz" instead of the number.

````python
# Loop through every number between 1 and 100.

for num in range(1,101):
    
    # Check for divisibility by 3 and 5 first.
    if (num % 3 == 0) and (num % 5 == 0):
        print('FizzBuzz')
        
    # If it's only divisible by 3, print Fizz.
    elif num % 3 == 0:
        print('Fizz')
        
    # If it's only divisible by 5, print Buzz.
    elif num % 5 == 0:
        print('Buzz')
        
    # Otherwise, print out the number.
    else:
        print(num)
````

## Functions
Code can get long and complicated. Functions allow us to "bottle up" large/complex pieces of code. We'll be able to re-use this later.

The general form of a function is like this:

```python
def function_name(arg1, arg2):
    # do some stuff
    return something
```

#### Example 1: Geometry!
Given a triangle's base and height, what's its area?

$$\text{area of triangle} = \frac{1}{2} * \text{base of triangle} * \text{height of triangle}$$

````python
# Define our function by writing def, 
# giving it a name, and specifying arguments.
def area_triangle(b, h):
    
    # Calculate the area of a triangle.
    area = (1/2) * b * h
    
    # We want to *return* the value of our output!
    return area
````

When defining a function, it's always a good idea to test it.

````python
# What is the area of a triangle with base 10 and height 10?
area_triangle(10, 10)
````

````python
# What is the area of a triangle with base 0.001 and height 2?
area_triangle(0.001, 2)
````

#### Example 2: Geometry Again!
Given a circle's radius, what's its area?

$$\text{area of circle} = \pi * \text{radius} ^ 2$$

````python
# We'll need a little help here.
import math

# Define a function called area_circle that takes one argument, radius.
def area_circle(radius):
    
    # Calculate the area using math.pi, our formula,
    # and return.
    return math.pi * radius ** 2
````

````python
# What is the area of a circle with radius 100?
area_circle(100)
````

````python
# What is the area of a circle with radius 0.013?
area_circle(0.013)
````

#### Example 4: Mean Function
Given a list of numbers, find their mean.

````python
# Define a function called my_mean that takes one argument, nums.
# You may assume nums is a list of numbers.

def my_mean(nums):
    # The mean is the sum of our numbers divided by
    # how many numbers we have.
    
    # Calculate the sum of our list.
    total = sum(nums)
    
    # Count up how many items we have in our list.
    n = len(nums)
    
    # Return the mean.
    return total / n
````

````python
# What's the mean of 1, 3, and 4?
my_mean([1, 3, 4])
````

````python
# What's the mean of 85, 92, 97, and 100?
my_mean([85, 92, 97, 100])
````

#### Example 5: Initialize

We want to extract someone's initials from their first and last names. For example, if our inputs are "Matt" and "Brems", our output should be "M.B."

````python
# Define a function called get_initials that takes two arguments:
# first and last. You may assume these are both strings.
def get_initials(first, last):
    
    # Take the first letter from the first name.
    # Capitalize it.
    first_init = first[0].upper()
    
    # Do the same with the last name.
    last_init = last[0].upper()
    
    # Return the first initial, a period, the last initial, and a period.
    return f'{first_init}.{last_init}.'
    # You could also return first_init + "." + last_init + "."
````

````python
# What are Matt Brems' initials?
get_initials('Matt', 'Brems')
````

````python
# What are Saad Usmani's initials?
get_initials('Saad', 'Usmani')
````

### Vowel Counting Function
That vowel counting problem was a doozy! Good thing we only had to solve it once. Let's bottle it up into a function so we can use it whenever we want!

````python
# Define a function called count_vowels that accepts one
# argument called "input_string" that is a string.

def count_vowels(input_string):
    
    # Instantiate counter at 0.
    n_vowels = 0

    # Iterate through our string.
    # Note that we use .lower() to account for upper-
    # case and lower-case vowels!
    for char in input_string.lower():
    
        # Check to see if the letter in our string
        # is a vowel.
        if char in 'aeiou':
        
            # If that's the case, then add 1 to our count.
            n_vowels += 1
        
    # RETURN the number of vowels.
    return n_vowels
````

````python
# Test it: How many vowels are in 'hello world?'
count_vowels('hello world')
````

### You Try: Pig Latin (another tough one!)
Let's write a function that converts **one single word** into Pig Latin. The (simplified) rules of Pig Latin are as follows:

> Take the first letter of a word and move it to the end and add _-ay_ as a suffix. If that letter was a vowel, add _-way_ instead. Only do this for words that are 3 letters or more.

You may assume the whole word is lowercase.

````python
# Define function called to_pig_latin_word that takes
# one argument, word.

def to_pig_latin_word(word):
    
    # First, check to see if the word is too short. If word
    # has 2 or fewer characters, just return the word.
    if len(word) <= 2:
        return word
    
    # Next, what happens if we have three or more letters?
    # We move the first letter of the word to the end.
    else:
        first_letter = word[0]
        
        # Then we add '-ay' or '-way' to the word. 
        # If our first letter was a vowel, add 'way.'
        if first_letter in 'aeiou':
            suffix = 'way'
        # Otherwise, add 'ay.'
        else:
            suffix = 'ay'
        
        # Combine our word together. This is:
        # our original word without the first letter,
        # THEN we add the first letter to the end,
        # THEN we add the suffix on at the end.
        return word[1:] + first_letter + suffix
````

````python
# How would you say "python" in Pig Latin?
to_pig_latin_word('python')
````

````python
# How would you say "avocado" in Pig Latin?
to_pig_latin_word('avocado')
````

````python
# How would you say "is" in Pig Latin?
to_pig_latin_word('is')
````

## Next:
Write a function that converts **an entire sentence** to Pig Latin. You may assume the whole sentence is lowercase and that there is no punctuation.

````python
# Define function called to_pig_latin that takes
# one argument, sentence.

def to_pig_latin(sentence):
    # Convert sentence from one string to a list of words.
    # The .split(' ') function will split our string
    # every time it sees a space.
    words = sentence.split(' ')
    
    # Create an empty list to store our "Pig Latin"-ified words.
    pig_words = []
    
    # Iterate through words.
    for word in words:
        
        # Convert each word to Pig Latin using our earlier function!
        # Then, add the "Pig Latin"-ified word to our list.
        pig_words.append(to_pig_latin_word(word))
        
    # Combine our list back together in one string, and return!
    return ' '.join(pig_words)
````

````python
# How would you say 'so long and thanks for all the fish' in Pig Latin?
to_pig_latin('so long and thanks for all the fish')
````

## Variable Scope
The "scope" of a variable is where it can be accessed. 
- If you define a variable inside of a function, it can only be accessed inside that function.
- If you define a variable outside of a function, it can be accessed either inside or outside of a function.

_(There are ways to bypass these rules, but they generally should not be used!)_

````python
# Define out_var outside our function.
out_var = "This variable was defined outside of a function!"
````

````python
# Define a function
def print_out_var():
    # out_var is accessible inside our function.
    print(out_var)
    # Define in_var inside our function.
    in_var = "This variable was defined inside of a function!"
````

````python
# Call print_out_var().
print_out_var()

# We defined out_var outside of a function,
# but out_var still works in a function.
````

````python
# Print in_var.
print(in_var) # <-- Error! in_var is not defined.

# We defined in_var inside of a function, BUT
# in_var is NOT accessible outside our function.
````

If you define a variable inside a function, it only lives there — unless you specify otherwise.
